# ARITHMETIC OPERATORS

```python
x = 10
y = 20

sum = x + y
print(sum)
```

```python
sub = x - y
print(sub)
```

```python
mult = x * y
print(mult)
```

```python
div = x / y
print(div)
```

```python
a = 7
b = 2

floordiv = a // b
print(floordiv)
```

```python
mod = a % b
print(mod)
```

```python
# a power b 7

exp = a ** b
print(exp)
```
