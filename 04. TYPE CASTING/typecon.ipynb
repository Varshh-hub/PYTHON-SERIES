# explicit type conversion

## to int()

```python
a = 67.90 # float
b = "42" # string
c = True # boolean
```

```python
print(a)
print(b)
print(c)
```

```python
print(type(a))
print(type(b))
print(type(c))
```

```python
a = int(a)
a
```

```python
b = int(b)
b
```

```python
c = int(c)
c # true = 1, false = 0
```

```python
x = "happy birthday"
x = int(x) #error
```

```python
x = "3.14"
x = int(x)
```
