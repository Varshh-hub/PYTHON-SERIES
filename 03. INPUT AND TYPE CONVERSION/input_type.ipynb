# simple calculator

1. get input - a , b

2. add - variable
3. sub -  variable
4. mult - variable
5. div - variable
6. fdiv - variable
7. mod - variable
8. exp - variable

9. print everything!

```python
a = int(input("enter a number : "))
b = int(input("enter another number: "))
```

```python
print(a)
print(b)
```

```python
add = a + b
sub = a - b
mult = a * b
div = a / b
fdiv = a // b
mod = a % b
exp = a ** b
```

```python
print("Addition :", add)
print("Subtraction :", sub)
print("Multiplication :", mult)
print("Division :", div)
print("Floor Division :", fdiv)
print("Modulus :", mod)
print("Exponentiation :", exp)
```
